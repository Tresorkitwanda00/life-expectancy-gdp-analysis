# Analyse de l'espérance de vie et du PIB
## Cadre du projet
### Objectif
Analyser la relation entre le produit intérieur brut (PIB) et l'espérance de vie dans six pays afin de déterminer si un PIB plus élevé est associé à une espérance de vie plus élevée.
### Questions de recherche
1. Comment l'espérance de vie a-t-elle évolué au fil du temps dans chaque pays ?
2. Comment le PIB a-t-il évolué au fil du temps dans chaque pays ?
3. Existe-t-il une relation entre le PIB et l'espérance de vie ?
4. La relation entre le PIB et l'espérance de vie varie-t-elle selon les pays ?
### Hypothèse
Les pays ayant un PIB plus élevé tendent à avoir une espérance de vie plus élevée.
### Données
Le jeu de données contient des informations sur le PIB et l'espérance de vie de six pays sur plusieurs années. Les données seront explorées, préparées, analysées et visualisées à l'aide de Python.

```python
df.groupby("Country")["Life expectancy at birth (years)"].mean().sort_values()
```

Germany présente l'espérance de vie moyenne la plus élevée, avec environ 79,66 ans.

Chile arrive en deuxième position, avec 78,94 ans.

Zimbabwe présente la moyenne la plus faible, avec environ 50,09 ans.

Les quatre autres pays ont des moyennes comprises entre 74,26 et 79,66 ans.

L'écart entre l'Allemagne et le Zimbabwe est d'environ 29,56 ans. Cela révèle des différences importantes entre les pays, sans toutefois expliquer à lui seul leurs causes.

```python
df.groupby("Country")["GDP"].mean().sort_values()
```

Les États-Unis possèdent le PIB moyen le plus élevé : environ 14075 milliards.

La Chine arrive en deuxième position, avec environ 4958 milliards.

Le Zimbabwe présente le PIB moyen le plus faible, avec environ 9,06 milliards.

L'Allemagne a un PIB moyen supérieur à celui du Mexique et du Chili.

Attention : le PIB total d'un pays ne mesure pas directement le niveau de vie de chaque habitant. Pour approfondir cette question, on pourrait notamment étudier le PIB par habitant.

2. Que pouvons-nous conclure pour le moment ?

Les résultats montrent que les pays ayant un PIB moyen élevé ne présentent pas nécessairement l'espérance de vie moyenne la plus élevée.

Par exemple, les États-Unis ont le PIB moyen le plus important, mais l'Allemagne présente l'espérance de vie moyenne la plus élevée parmi les six pays. Cela suggère que la relation entre le PIB et l'espérance de vie n'est pas parfaitement proportionnelle.

Il faut toutefois étudier les données plus directement avant de conclure à l'existence, à la force ou au sens d'une corrélation.
